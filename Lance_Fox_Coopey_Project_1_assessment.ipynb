## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python
def validate_read(sequence, reference):
    """Check that the read being looked at is valid compared to the reference.

    Valid reads will only have A,T,G or C, be the same length as the reference
    and lowercase letters will be updated to uppercase.

    Args:
        sequence (str): the read we want to check.
        reference (str): the reference sequence we are comparing against.

    Returns:
        dict: "sequence" (the upper-case read), "valid" (True or False)
        and "errors" (a list of error codes, empty if the read is valid).
    """
    # Making any lower case text uppercase.
    sequence = sequence.upper()
    # Creating errors list.
    errors = []
    # Making sure that the length matches the refrence.
    if len(sequence) != len(reference):
        errors.append("wrong_length")
    # Now making sure it is in the group A,T,G or C.
    for base in sequence:
        if base not in "ACGT":
            errors.append("invalid_characters")
            break # So it doesnt crash.
    # Making sure it is only valid if there are no errors.
    valid = len(errors) == 0

    return {"sequence": sequence, "valid": valid, "errors": errors}
````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python
def find_mutations(sequence, reference):
    """Find all positions where read is diffrent than the refrence.
    Positions are numbered from 1 and are assumed to be valid due to Step 1.
    Args:
        sequence (str): the read to check.
        reference (str): the reference sequence to compare against.

    Returns:
        list: one dictionary per mismatch, in increasing position order, each
        with the keys "position", "reference_base" and "read_base".
        An empty list means the read completely matches the reference.
    """
    # Creating empty mutation list
    mutations = []
    # Pairing base by base, adding a counter to each one so it doesnt start at 0
    for position, (read_base, reference_base) in enumerate(zip(sequence, reference), start=1):
        # Chacking if the two bases are diffrent
        if read_base != reference_base:
            # If there is a diffrence then mutations is appended with a dictionary
            mutations.append({
                "position": position,
                "reference_base": reference_base,
                "read_base": read_base,
            })
    return mutations
````

### 💣 Test your code

````python
# Using the same test refrence as before
test_reference = "ATGC"

# Normal tests
test_mutations = find_mutations("GTGA", test_reference)
assert test_mutations == [{"position": 1, "reference_base": "A", "read_base": "G"},
                          {"position": 4, "reference_base": "C", "read_base": "A"}]
assert find_mutations("ATGC", test_reference) == [] # No mismatch
assert find_mutations("ACGC", test_reference) == [{"position": 2, "reference_base": "T", "read_base": "C"}]  # One mismatch in the middle

# Edge cases
assert find_mutations("CTGC", test_reference) == [{"position": 1, "reference_base": "A", "read_base": "C"}]   # First position
assert find_mutations("ATGA", test_reference) == [{"position": 4, "reference_base": "C", "read_base": "A"}]   # Last position
assert find_mutations("TACG", test_reference) == [                                                            # Every base mismatched, in order
    {"position": 1, "reference_base": "A", "read_base": "T"},
    {"position": 2, "reference_base": "T", "read_base": "A"},
    {"position": 3, "reference_base": "G", "read_base": "C"},
    {"position": 4, "reference_base": "C", "read_base": "G"},
]
assert find_mutations("", "") == []   # Empty sequences have nothing to compare

print(test_mutations)
print("All find_mutations tests have been passed")
````

## Step 3 — Analyse all reads
## Function: `analyse_reads(reads, reference)`

### 👍 You can assume:
- `reads` is a list of dictionaries containing keys `read_id`, `sample`, `group` and `sequence`, each with string values.
- `reference` is a non-empty string containing only A/C/G/T.

### 🎯 Aims:
- Use `validate_read` and `find_mutations` to return a **new list of dictionaries containing information about all reads** (even invalid ones, one dictionary per read), preserving the input read order.
- Each read dictionary should contain the following keys (values below are examples):
```python
{
    "read_id": "read_0001",
    "sample": "sample1",
    "group": "treatment",
    "sequence": "GTGA",             # normalised upper-case sequence
    "status": "valid",              # "valid" or "invalid"
    "errors": [],                   # validation error codes
    "mutations": [
        {"position": 1, "reference_base": "A", "read_base": "G"},
        {"position": 4, "reference_base": "C", "read_base": "A"}
    ]
}
```
- Do not change the original read dictionaries.

### 🔎 Expected outputs
- Valid reads have `status="valid"` and `mutations` listing all mutations in the dictionary format shown above.
- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python
def analyse_reads(reads, reference):
    """Analyse each read by validating it and finding its mutations.

    Each read is checked with validate_read,
    Valid reads are compared to the reference with find_mutations;
    Invalid reads are kept in the output but get an empty mutation list.

    Args:
        reads (list): Dictionaries with the keys, all with string values.
        reference (str): The reference sequence.

    Returns:
        list: A new list with one dictionary per read, in the same order as
        the input. Each of which contains
        "read_id", "sample", "group", "sequence" (upper
        case), "status" ("valid" or "invalid"), "errors" (list of error
        codes) and "mutations" (list of mutation dictionaries). And an empty
        input gives an empty list.
    """
    # Creating a new list
    analysed_reads = []

    for read in reads:
        validation = validate_read(read["sequence"], reference) #Checks the read and saves the result

        if validation["valid"]:
            status = "valid"
            mutations = find_mutations(validation["sequence"], reference) # If the read is valid, its is marked “valid” and find its mutations.
        else:
            status = "invalid"
            mutations = []  # If not it is marked invalid and given no mutation

        analysed_reads.append({     # Adds a new dictionary for this read to the results
            "read_id": read["read_id"],
            "sample": read["sample"],
            "group": read["group"],
            "sequence": validation["sequence"],
            "status": status,
            "errors": list(validation["errors"]),
            "mutations": mutations,
        })

    return analysed_reads
````

- How to test your function (e.g. using test datasets and `assert`) to make sure it can handle any input (provided in the list of dictionaries format output by `analyse_reads`)

### 💻 Enter your code here

````python
def get_top_positions(position_counts):
    """Return up to 3 positions with the most mismatch events."""
    ranked = sorted(position_counts, key=lambda pos: (-position_counts[pos], pos)) #Sorting by count and then position
    return ranked[:3] # Gives fewer than three if there are not enough positions.


def build_summary(analysed_reads):
    """Count mismatches for each sample, using only valid reads only for mismatches.

    Args:
        analysed_reads (list): dictionaries in the format given by analyse_reads.

    Returns:
        dict: one dictionary per sample.
        Empty input gives {}.
    """
    summary = {} #Epty dictionary
    for read in analysed_reads:
        # The first time sample is seen it gets a new record with all of its counts at 0
        info = summary.setdefault(read["sample"], {
            "group": read["group"], "valid_reads": 0, "invalid_reads": 0,
            "mismatch_reads": 0, "mismatch_percentage": None, "mismatch_events": 0,
            "substitution_counts": {}, "position_counts": {}, "top_positions": []})
        # Any invalid reads are only counted then skipped, not adding to mismatches
        if read["status"] == "invalid":
            info["invalid_reads"] += 1
            continue

        info["valid_reads"] += 1
        # A read with any mismatches is counted as one mismatch
        if read["mutations"]:
            info["mismatch_reads"] += 1
        # Each mutation is one mismatch event
        for mutation in read["mutations"]:
            info["mismatch_events"] += 1
            sub = mutation["reference_base"] + ">" + mutation["read_base"]
            pos = mutation["position"]                                     # Making a subsitution label and getting its position
            info["substitution_counts"][sub] = info["substitution_counts"].get(sub, 0) + 1
            info["position_counts"][pos] = info["position_counts"].get(pos, 0) + 1    # Adds 1 to counts and starts at 0 if not seen yet

    # Percentage and top positions for every read
    for info in summary.values():
        if info["valid_reads"] > 0: # Has to be at least 1 read
            info["mismatch_percentage"] = info["mismatch_reads"] * 100 / info["valid_reads"]
        info["top_positions"] = get_top_positions(info["position_counts"])
    return summary


def summarise_samples(analysed_reads):
    """Summarise mismatches per sample and print a section for each group.

    Args:
        analysed_reads (list): dictionaries in the format given by analyse_reads.

    Returns:
        dict: one nested dictionary per sample.
    """
    summary = build_summary(analysed_reads)

    for group in sorted({info["group"] for info in summary.values()}):     # Finding samples in this group and sorting so they are in the same order
        samples = sorted(s for s in summary if summary[s]["group"] == group)
        print(f"\n===== GROUP: {group} =====")

        # Table of summarised results
        print(f"{'Sample':<10}{'Valid':>7}{'Invalid':>9}{'% mismatch':>12}{'Events':>8}   Top 3 positions")
        for s in samples:
            info = summary[s]
            pct = "N/A" if info["mismatch_percentage"] is None else f"{info['mismatch_percentage']:.2f}"  #No percentage for no valid reads
            top = ", ".join(map(str, info["top_positions"])) or "None"  #Top positions or none is there are no mismatches
            print(f"{s:<10}{info['valid_reads']:>7}{info['invalid_reads']:>9}{pct:>12}{info['mismatch_events']:>8}   {top}") # One row per sample
            if pct == "N/A":
                print(f"  Note: {s} has no valid reads, so percentages cannot be calculated.")

        # Table of substitution mutations: count (% per 100 valid reads)
        print(f"\n{'Substitution':<14}" + "".join(f"{s:>18}" for s in samples))
        for sub in sorted({k for s in samples for k in summary[s]["substitution_counts"]}): # Showing substitutions as A → G instead of A>G
            line = f"{sub.replace('>', ' → '):<14}"
            for s in samples:
                count = summary[s]["substitution_counts"].get(sub, 0)
                valid = summary[s]["valid_reads"]
                line += f"{count} ({count * 100 / valid:.2f}%)".rjust(18) if valid else f"{count} (N/A)".rjust(18) # Count and percentage per 100 valid reads
            print(line)
    return summary
````

## Step 5 — Exploratory comparison of the top mismatch positions

## Function:
`compare_positions(sample_summary, control_group="control", treated_group="treatment")`

### 👍 You can assume

- `sample_summary` is a dictionary in the format returned by the `summarise_samples` function
- `control_group` and `treated_group` are strings
- All samples have been analysed against the same reference sequence.

### 🎯 Aims

- Using the `position_counts` returned by `summarise_samples`, sum the mismatch counts at every position, **combining samples from `control_group` and `treatment_group`** (you should end up with a 1D array/list with the same length as a read).
- Select the **top 3 positions with the highest combined counts** (or fewer if there are less than three mismatches), choosing the earlier reference position in the case of a tie.
- For each position selected, calculate **mismatch percentages** separately for each biological sample.
- Compare treatment groups using the supplied statistical function `calculate_statistics` (see below).
- **Print out** the results of the statistical comparison clearly in standard scientific style, including appropriate statistical details based on the test used (a two-sided Welch's t-test).

🛠 **Parameters:**

- `sample_summary`: output from `summarise_samples` function.
- `control_group` and `treated_group`: labels identifying the groups to compare.

### 🧮 Sample-level calculations

For each selected position and sample, calculate:

**Mismatch percentage = valid reads with a mismatch at that position ÷ total valid reads × 100**

Count any substitution at the selected position. Include samples with zero mismatches, but exclude samples with no valid reads and report their exclusion.

Reuse helper functions from Step 4 where appropriate to reduce code complexity and repetition.

### 📊 Statistical comparison

For each position, pass the sample percentages to the supplied function as a list (one per group) and store the result, e.g.:

```python
statistical_result = calculate_statistics(control_percentages, treated_percentages)
```

Each list must contain **one percentage per biological sample**.

The supplied function returns:

```python
{"statistic": float, "p_value": float}
```

It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````

### 📋 Return format

Return a list with one dictionary per selected position:

```python
[
    {
        "position": 42,
        "control_n": 3,
        "treated_n": 3,
        "control_mean": 0.5,
        "treated_mean": 2.0,
        "difference": 1.5,
        "statistic": ...,
        "p_value": ...
    }
]
```

Means are sample mismatch percentages; `difference` is **treated mean − control mean**, in percentage points. Use `None` for results that cannot be calculated.

Print **one brief line per position**, reporting group means, their difference, sample numbers and p-value.

### 💻 Enter your code here

````python
def get_group_samples(sample_summary, group):
    """Split the samples into two groups: valid reads and those without."""
    included = sorted(s for s, info in sample_summary.items()
                      if info["group"] == group and info["valid_reads"] > 0)     # Samples in this group have at least 1 valid read
    excluded = sorted(s for s, info in sample_summary.items()
                      if info["group"] == group and info["valid_reads"] == 0)    # No valid reads
    return included, excluded


def combine_position_counts(sample_summary, groups):
    """Add up the mismatch events at each position across all samples in the given groups."""
    combined = {}                                                # Empty dictionary of position for total mismatch events
    for info in sample_summary.values():
        if info["group"] in groups:                              # Only using samples from the groups being compared
            for pos, count in info["position_counts"].items():
                combined[pos] = combined.get(pos, 0) + count     # Adds to the total
    return combined


def calculate_position_percentages(sample_summary, samples, position):
    """Return the mismatch percentage at one position for each sample."""
    percentages = []                                             # One percentage per sample
    for s in samples:
        mismatches = sample_summary[s]["position_counts"].get(position, 0)   # 0 if this sample has no mismatch at this position
        valid = sample_summary[s]["valid_reads"]                 # Total valid reads in this sample
        percentages.append(mismatches * 100 / valid)             # Calculating Mismatch %
    return percentages


def calculate_mean(values):
    """Return the mean, if none then list is empty."""
    return sum(values) / len(values) if values else None


def format_p_value(p_value):
    """Format a p-value."""
    if p_value is None:
        return "p = N/A"
    if p_value < 0.001:                                          # Very small p-values are shown as p < 0.001
        return "p < 0.001"
    return f"p = {p_value:.3f}"                                  # If larger then its put to 3 decimal places


def format_value(value, template):
    """Format a number using the template."""
    return "N/A" if value is None else template.format(value)    # None becomes N/A instead of crashing


def format_result_line(result, control_group, treated_group):
    """One brief line reporting the means, difference, sample numbers and p-value."""
    control_mean = format_value(result["control_mean"], "{:.2f}")
    treated_mean = format_value(result["treated_mean"], "{:.2f}")
    difference = format_value(result["difference"], "{:+.2f}")         # + or - sign shows the direction
    statistic = format_value(result["statistic"], "{:.2f}")            # The t statistic
    return (f"Position {result['position']}: {control_group} mean {control_mean}% "
            f"vs {treated_group} mean {treated_mean}% "                # Group means
            f"(difference = {difference} percentage points; "          # Treated mean - control mean
            f"n = {result['control_n']} vs {result['treated_n']}; "    # Number of samples
            f"t = {statistic}, {format_p_value(result['p_value'])})")  # Statistic and p-value


def compare_positions(sample_summary, control_group="control", treated_group="treatment"):
    """Compare mismatch percentages at the top 3 positions between two groups.

    The mismatch counts at each position are combined across both groups,
    and the top 3 positions are chosen.
    For each one,
    the percentage per sample is compared between groups with calculate_statistics,
    and one line is printed per position.

    Args:
        sample_summary (dict): output from summarise_samples.
        control_group (str): label of the control group.
        treated_group (str): label of the treated group.

    Returns:
        list: one dictionary per selected position, with the keys "position",
        "control_n", "treated_n", "control_mean", "treated_mean" (sample mismatch
        percentages), "difference" (treated mean - control mean, in percentage
        points), "statistic" and "p_value". Values that cannot be calculated
        are None. Empty input gives [].
    """
    if control_group == treated_group:                           # Comparing a group with itself
        raise ValueError("control_group and treated_group must be different")

    control_samples, control_excluded = get_group_samples(sample_summary, control_group)   # Usable and excluded control samples
    treated_samples, treated_excluded = get_group_samples(sample_summary, treated_group)   # Usable and excluded treated samples
    for s in control_excluded + treated_excluded:                # Reporting any samples that are not included
        print(f"Note: {s} has no valid reads so it was excluded from the comparison.")

    combined_counts = combine_position_counts(sample_summary, [control_group, treated_group])   # Total events per position for both groups
    top_positions = get_top_positions(combined_counts)           # Top 3 positions by reusing Step 4

    results = []                                                 # One dictionary for each selected position
    for position in top_positions:
        control_percentages = calculate_position_percentages(sample_summary, control_samples, position)   # One % per control sample
        treated_percentages = calculate_position_percentages(sample_summary, treated_samples, position)   # One % per treated sample

        control_mean = calculate_mean(control_percentages)       # Mean of the control sample percentages
        treated_mean = calculate_mean(treated_percentages)       # Mean of the treated sample percentages
        difference = None if control_mean is None or treated_mean is None else treated_mean - control_mean   # Treated - control and none if either group is empty

        statistical_result = calculate_statistics(control_percentages, treated_percentages)   # Welch's t-test

        results.append({                                         # Storing the result
            "position": position,
            "control_n": len(control_percentages),
            "treated_n": len(treated_percentages),
            "control_mean": control_mean,
            "treated_mean": treated_mean,
            "difference": difference,
            "statistic": statistical_result["statistic"],
            "p_value": statistical_result["p_value"],
        })

    print(f"\nTop mismatch positions: {control_group} vs {treated_group} (Welch's t-test, two-sided)")
    for result in results:
        print(format_result_line(result, control_group, treated_group))    # One line per position
    return results
````
